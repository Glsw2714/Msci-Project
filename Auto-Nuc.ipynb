```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nucml.datasets as nuc_data
from autogluon.tabular import TabularPredictor

# --- CONFIGURATION ---
CUTOFF_LOG10_EV = 5.0   # 100 keV cutoff
TARGET_MT = 18          # 18 = (n,f), 16 = (n,2n)
LIBRARY = "endfb8.0"

# --- PHYSICS HELPER FUNCTIONS ---
def get_physics_features(df):
    """Adds basic nuclear physics features to the dataframe."""
    df['N'] = df['A'] - df['Z']             # Neutron number
    df['N_Z_Ratio'] = df['N'] / df['Z']     # Stability ratio
    # Magic number proximity (simplified for actinides)
    df['Dist_Magic_N'] = abs(df['N'] - 126) 
    df['Coulomb_Term'] = (df['Z']**2) / (df['A']**(1/3))
    return df

# --- DATA SELECTION ---
# Focusing on Actinides for bracketing
training_isotopes = [
    {'Z': 90, 'A': 232}, # Th-232
    {'Z': 90, 'A': 229}, # Th-229 (Corrected from Z=70)
    {'Z': 92, 'A': 233}, # U-233
    {'Z': 92, 'A': 235}, # U-235
    {'Z': 92, 'A': 238}, # U-238
    {'Z': 95, 'A': 241}, # Am-241
    {'Z': 95, 'A': 243}, # Am-243
    {'Z': 96, 'A': 242}, # Cm-242
    {'Z': 96, 'A': 244}, # Cm-244
    {'Z': 96, 'A': 245}, # Cm-245
]

dfs = []
print(f"Loading MT={TARGET_MT} from {LIBRARY}...")

for iso in training_isotopes:
    try:
        # 1. Resolve Element Symbol
        # Note: In a full project, build a robust Z->Symbol mapper
        symbol = list(nuc_data.elements_dict.keys())[list(nuc_data.elements_dict.values()).index(iso['Z'])]
        element_id = f"{symbol}{iso['A']}"
        
        # 2. Load Data
        df = nuc_data.load_evaluation(element_id, TARGET_MT, library=LIBRARY)
        
        # 3. Log-Log Transformation (CRITICAL for XS)
        # NucML usually loads 'Energy' in eV and 'Data' in barns (linear)
        # We filter FIRST, then log transform
        df['Log_Energy'] = np.log10(df['Energy'])
        df_fast = df[df['Log_Energy'] > CUTOFF_LOG10_EV].copy()
        
        if not df_fast.empty:
            # Handle Zero/Negative XS before logging
            df_fast = df_fast[df_fast['Data'] > 0].copy()
            df_fast['Log_XS'] = np.log10(df_fast['Data'])
            
            # 4. Feature Engineering
            df_fast['Z'] = iso['Z']
            df_fast['A'] = iso['A']
            df_fast['Is_Odd_A'] = int(iso['A'] % 2 != 0)
            df_fast = get_physics_features(df_fast)
            
            dfs.append(df_fast)
            print(f" -> {element_id}: Loaded {len(df_fast)} rows")
        else:
            print(f" -> {element_id}: Dropped (Empty or below cutoff)")
            
    except Exception as e:
        print(f" -> Skipped {iso['Z']}-{iso['A']}: Data not found in library")

# --- TRAINING ---
train_fast = pd.concat(dfs, ignore_index=True)

# We drop the linear columns so the model doesn't cheat or get confused
train_data = train_fast.drop(columns=['Energy', 'Data'])

print("Training AutoGluon...")
# Using 'mean_absolute_error' on Log data approximates MAPE on linear data
predictor = TabularPredictor(label='Log_XS', eval_metric='mean_absolute_error').fit(
    train_data=train_data,
    presets='medium_quality', # 'best' takes a long time, use medium for dev
    time_limit=600
)

# --- TESTING (Pu-239) ---
print("Testing on Pu-239...")
test_df_raw = nuc_data.load_evaluation("Pu239", TARGET_MT, library=LIBRARY)
test_df_raw['Log_Energy'] = np.log10(test_df_raw['Energy'])

# Filter and Prep
test_fast = test_df_raw[test_df_raw['Log_Energy'] > CUTOFF_LOG10_EV].copy()
test_fast = test_fast[test_fast['Data'] > 0].copy() # Remove zeros
test_fast['Log_XS_Actual'] = np.log10(test_fast['Data'])

# Inject Features for Pu-239
test_fast['Z'] = 94
test_fast['A'] = 239
test_fast['Is_Odd_A'] = 1
test_fast = get_physics_features(test_fast)

# Predict
test_fast['Log_XS_Pred'] = predictor.predict(test_fast)

# --- PLOTTING ---
plt.figure(figsize=(10, 6))
# Plotting in Log-Log space (Linear visual of the Log values)
plt.plot(test_fast['Log_Energy'], test_fast['Log_XS_Actual'], label='ENDF/B-VIII.0 (Actual)', color='black', linewidth=1.5)
plt.plot(test_fast['Log_Energy'], test_fast['Log_XS_Pred'], label='AutoGluon (Predicted)', color='red', linestyle='--', linewidth=1.5)

plt.title(f"Prediction of Pu-239 (n,f) based on Th, U, Am, Cm")
plt.xlabel("Log10 Energy (eV)")
plt.ylabel("Log10 Cross Section (barns)")
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()
```
