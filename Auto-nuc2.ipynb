```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nucml.datasets as nuc_data
from autogluon.tabular import TabularPredictor

# --- CONFIGURATION ---
TARGET_ISO = {'Z': 94, 'A': 239}  # Pu-239 (Target)
REACTION_MT = 16                  # (n,2n) Reaction
FAST_THRESHOLD_EV = 6e6           # 6 MeV (Physical threshold for n,2n)

# Expanded Training List (U-series, Pu-neighbors, Minor Actinides)
training_isotopes = [
    {'Z': 90, 'A': 232}, # Th-232
    {'Z': 92, 'A': 233}, {'Z': 92, 'A': 234}, {'Z': 92, 'A': 235}, 
    {'Z': 92, 'A': 236}, {'Z': 92, 'A': 238}, # Uranium Series
    {'Z': 93, 'A': 237}, # Np-237
    {'Z': 94, 'A': 238}, {'Z': 94, 'A': 240}, 
    {'Z': 94, 'A': 241}, {'Z': 94, 'A': 242}, # Pu Neighbors
    {'Z': 95, 'A': 241}, {'Z': 95, 'A': 243}, # Am Isotopes
]

# --- DATA LOADING HELPER ---
def load_isotope_data(z, a, mt, energy_cutoff=None):
    try:
        symbol = list(nuc_data.elements_dict.keys())[list(nuc_data.elements_dict.values()).index(z)]
        element_id = f"{symbol}{a}"
        
        # Load Data (Defaults to Log10 in many NucML configs)
        df = nuc_data.load_evaluation(element_id, mt, library="endfb8.0")
        
        # Add Physics Features
        df['Z'] = z
        df['A'] = a
        df['Is_Odd_A'] = int(a % 2 != 0)
        
        # Apply Energy Cutoff if specified
        if energy_cutoff is not None:
            df = df[df['Energy'] > energy_cutoff].copy()
            
        return df, element_id
    except Exception as e:
        return pd.DataFrame(), str(e)

# ==========================================
# 1. LOAD & CLEAN DATA
# ==========================================
print(f"Loading Training Data for {len(training_isotopes)} isotopes...")

# Load Whole Spectrum Data
dfs_whole = []
for iso in training_isotopes:
    df, name = load_isotope_data(iso['Z'], iso['A'], REACTION_MT)
    if not df.empty:
        dfs_whole.append(df)
        print(f" -> Loaded {name}: {len(df)} rows")

train_data_whole = pd.concat(dfs_whole, ignore_index=True)

# --- CRITICAL FIX: HANDLE -INF (Log Zero) ---
# Replace -inf with -20.0 (effectively zero barns) to fix ValueError
train_data_whole['Data'] = train_data_whole['Data'].replace([np.inf, -np.inf], -20.0)
train_data_whole = train_data_whole.dropna(subset=['Data']) # Drop any remaining NaNs

# Load Target (Pu-239) for Testing
test_data_whole, _ = load_isotope_data(TARGET_ISO['Z'], TARGET_ISO['A'], REACTION_MT)
test_data_whole['Data'] = test_data_whole['Data'].replace([np.inf, -np.inf], -20.0)

print(f"\nTotal Training Rows (Whole): {len(train_data_whole)}")

# ==========================================
# 2. EXPERIMENT: WHOLE SPECTRUM TRAINING
# ==========================================
print("\n" + "="*40)
print("  RUN 1: WHOLE SPECTRUM TRAINING")
print("="*40)

predictor_whole = TabularPredictor(label='Data', eval_metric='r2', path="AutogluonModels/n2n_whole").fit(
    train_data=train_data_whole,
    time_limit=300, 
    presets='medium_quality' 
)

# Predict on full range
test_data_whole['Pred_Whole'] = predictor_whole.predict(test_data_whole)

# ==========================================
# 3. EXPERIMENT: FAST REGIME TRAINING
# ==========================================
print("\n" + "="*40)
print("  RUN 2: FAST REGIME TRAINING (> 6 MeV)")
print("="*40)

# Calculate Log10 Threshold (6 MeV = 6e6 eV -> Log10 ~ 6.78)
LOG_THRESHOLD = np.log10(FAST_THRESHOLD_EV)

# Filter Data for Fast Regime
train_data_fast = train_data_whole[train_data_whole['Energy'] > LOG_THRESHOLD].copy()
test_data_fast = test_data_whole[test_data_whole['Energy'] > LOG_THRESHOLD].copy()

print(f"Total Training Rows (Fast): {len(train_data_fast)}")

predictor_fast = TabularPredictor(label='Data', eval_metric='r2', path="AutogluonModels/n2n_fast").fit(
    train_data=train_data_fast,
    time_limit=300,
    presets='medium_quality'
)

# Predict on Fast range
test_data_fast['Pred_Fast'] = predictor_fast.predict(test_data_fast)

# ==========================================
# 4. VISUALIZATION
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Plot 1: Whole Spectrum (Log-Log to see threshold behavior)
# We plot in Linear Y-scale for cross-section to clearly see the "Zero" region
axes[0].plot(10**test_data_whole['Energy'], 10**test_data_whole['Data'], 'b-', label='Actual Pu-239', alpha=0.5, linewidth=2)
axes[0].plot(10**test_data_whole['Energy'], 10**test_data_whole['Pred_Whole'], 'r--', label='Model (Whole Spectrum)', alpha=0.8)
axes[0].axvline(x=FAST_THRESHOLD_EV, color='k', linestyle=':', label='6 MeV Threshold')
axes[0].set_title("(n,2n) Whole Spectrum: Learning the Threshold")
axes[0].set_xlabel("Energy (eV)")
axes[0].set_ylabel("Cross Section (barns)")
axes[0].set_xscale('log')
# axes[0].set_yscale('log') # Optional: Toggle log y to see low values better
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2: Fast Regime Only
axes[1].plot(10**test_data_fast['Energy'], 10**test_data_fast['Data'], 'b-', label='Actual Pu-239', alpha=0.5, linewidth=2)
axes[1].plot(10**test_data_fast['Energy'], 10**test_data_fast['Pred_Fast'], 'g--', label='Model (Fast Only)', alpha=0.8)
axes[1].set_title("(n,2n) Fast Regime: Shape Interpolation")
axes[1].set_xlabel("Energy (eV)")
axes[1].set_ylabel("Cross Section (barns)")
axes[1].set_xscale('log')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("n2n_Comparison_Fixed.png")
plt.show()

# Final Scores
score_whole = predictor_whole.evaluate(test_data_whole, silent=True)['r2']
score_fast = predictor_fast.evaluate(test_data_fast, silent=True)['r2']

print(f"\nFinal Test Scores (R2):")
print(f"Whole Spectrum Model: {score_whole:.4f}")
print(f"Fast Regime Model:    {score_fast:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nucml.datasets as nuc_data
from autogluon.tabular import TabularPredictor

# --- CONFIGURATION ---
# Target: Niobium-93 (Mid-mass, standard dosimetry material)
TARGET_ISO = {'Z': 41, 'A': 93} 
REACTION_MT = 16 # (n,2n) Reaction

# Threshold varies by mass, but ~8-10 MeV is common for this range.
# We will use 8 MeV as the cutoff for the "Fast Regime" experiment.
FAST_THRESHOLD_EV = 8e6 

# Wide Training Set (A=20 to A=200)
# A mix of structural, fission product, and heavy materials.
training_isotopes = [
    {'Z': 13, 'A': 27},  # Al-27
    {'Z': 22, 'A': 48},  # Ti-48
    {'Z': 24, 'A': 52},  # Cr-52
    {'Z': 26, 'A': 56},  # Fe-56
    {'Z': 28, 'A': 58},  # Ni-58
    {'Z': 29, 'A': 63},  # Cu-63
    {'Z': 40, 'A': 90},  # Zr-90
    {'Z': 42, 'A': 98},  # Mo-98
    {'Z': 47, 'A': 107}, # Ag-107
    {'Z': 48, 'A': 114}, # Cd-114
    {'Z': 50, 'A': 120}, # Sn-120
    {'Z': 56, 'A': 138}, # Ba-138
    {'Z': 63, 'A': 151}, # Eu-151
    {'Z': 64, 'A': 157}, # Gd-157
    {'Z': 74, 'A': 184}, # W-184
    {'Z': 79, 'A': 197}, # Au-197
]

# --- DATA LOADING HELPER ---
def load_isotope_data(z, a, mt, energy_cutoff=None):
    try:
        # Get Symbol
        symbol = list(nuc_data.elements_dict.keys())[list(nuc_data.elements_dict.values()).index(z)]
        element_id = f"{symbol}{a}"
        
        # Load Data
        df = nuc_data.load_evaluation(element_id, mt, library="endfb8.0")
        
        # Add Features
        df['Z'] = z
        df['A'] = a
        df['Is_Odd_A'] = int(a % 2 != 0)
        
        # Cutoff
        if energy_cutoff is not None:
            df = df[df['Energy'] > energy_cutoff].copy()
            
        return df, element_id
    except Exception as e:
        return pd.DataFrame(), str(e)

# ==========================================
# 1. LOAD & CLEAN DATA
# ==========================================
print(f"Loading Training Data (A=20-200)...")

dfs_whole = []
for iso in training_isotopes:
    df, name = load_isotope_data(iso['Z'], iso['A'], REACTION_MT)
    if not df.empty:
        dfs_whole.append(df)
        print(f" -> Loaded {name}: {len(df)} rows")

train_data_whole = pd.concat(dfs_whole, ignore_index=True)

# Fix -inf (Log Zero)
train_data_whole['Data'] = train_data_whole['Data'].replace([np.inf, -np.inf], -20.0)
train_data_whole = train_data_whole.dropna(subset=['Data'])

# Load Target (Nb-93)
test_data_whole, target_name = load_isotope_data(TARGET_ISO['Z'], TARGET_ISO['A'], REACTION_MT)
test_data_whole['Data'] = test_data_whole['Data'].replace([np.inf, -np.inf], -20.0)

print(f"\nTotal Training Rows: {len(train_data_whole)}")
print(f"Target: {target_name} ({len(test_data_whole)} rows)")


# ==========================================
# 2. EXPERIMENT: WHOLE SPECTRUM
# ==========================================
print("\n" + "="*40)
print("  RUN 1: WHOLE SPECTRUM TRAINING")
print("="*40)

predictor_whole = TabularPredictor(label='Data', eval_metric='r2', path="AutogluonModels/n2n_medium_whole").fit(
    train_data=train_data_whole,
    time_limit=300, 
    presets='high_quality'
)
test_data_whole['Pred_Whole'] = predictor_whole.predict(test_data_whole)

# ==========================================
# 3. EXPERIMENT: FAST REGIME (> 8 MeV)
# ==========================================
print("\n" + "="*40)
print(f"  RUN 2: FAST REGIME TRAINING (> {FAST_THRESHOLD_EV/1e6} MeV)")
print("="*40)

# 8 MeV = 8e6 eV -> Log10 ~ 6.90
LOG_THRESHOLD = np.log10(FAST_THRESHOLD_EV)

# Filter Data
train_data_fast = train_data_whole[train_data_whole['Energy'] > LOG_THRESHOLD].copy()
test_data_fast = test_data_whole[test_data_whole['Energy'] > LOG_THRESHOLD].copy()

print(f"Fast Training Rows: {len(train_data_fast)}")

predictor_fast = TabularPredictor(label='Data', eval_metric='r2', path="AutogluonModels/n2n_medium_fast").fit(
    train_data=train_data_fast,
    time_limit=300,
    presets='medium_quality'
)
test_data_fast['Pred_Fast'] = predictor_fast.predict(test_data_fast)

# ==========================================
# 4. VISUALIZATION
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Plot 1: Whole
axes[0].plot(10**test_data_whole['Energy'], 10**test_data_whole['Data'], 'b-', label=f'Actual {target_name}', alpha=0.6, linewidth=2)
axes[0].plot(10**test_data_whole['Energy'], 10**test_data_whole['Pred_Whole'], 'r--', label='Model (Whole)', alpha=0.8)
axes[0].axvline(x=FAST_THRESHOLD_EV, color='k', linestyle=':', label='8 MeV Threshold')
axes[0].set_title("(n,2n) Whole Spectrum (A=20-200)")
axes[0].set_xlabel("Energy (eV)")
axes[0].set_ylabel("Cross Section (barns)")
axes[0].set_xscale('log')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2: Fast
axes[1].plot(10**test_data_fast['Energy'], 10**test_data_fast['Data'], 'b-', label=f'Actual {target_name}', alpha=0.6, linewidth=2)
axes[1].plot(10**test_data_fast['Energy'], 10**test_data_fast['Pred_Fast'], 'g--', label='Model (Fast Only)', alpha=0.8)
axes[1].set_title(f"(n,2n) Fast Regime (A=20-200)")
axes[1].set_xlabel("Energy (eV)")
axes[1].set_xscale('log')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("n2n_MediumMass_Comparison.png")
plt.show()

# Scores
print(f"\nFinal R2 Scores for {target_name}:")
print(f"Whole Spectrum Model: {predictor_whole.evaluate(test_data_whole, silent=True)['r2']:.4f}")
print(f"Fast Regime Model:    {predictor_fast.evaluate(test_data_fast, silent=True)['r2']:.4f}")
```
